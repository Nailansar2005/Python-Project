## Data Wrangling

4.Datanın sütunları ilə tanış olun. Onlar hansı data tiplərindədir? Bütün sütunların data tipləri uyğundur?

```python
df.info()
```

Xeyr, bütün sütunların tipi uyğun deyil: "As of" və "Netflix Release Date" sütunları tarix (datetime) olmalı ikən mətn (object/string) kimi saxlanılıb.

5.Düzgün data tipində olmayan müvafiq data tiplərinə keçirin

```python
df['As of'] = pd.to_datetime(df['As of'],errors='coerce')
df['Netflix Release Date'] = pd.to_datetime(df['Netflix Release Date'],errors='coerce')
df.info
```

6.Hansı sütunlarda boş sətrlər var? Hansı sütunlarda neçə boş sətr var?

```python
df.isnull().sum()
```

7.Boş sətrlər olan sütunda olan bütün boş məlumatları "No" yazısı ilə doldurun

```python
df = df.fillna(value = 'No')
df
```

## Analysis

8.Datada film tiplərinə görə paylanmanı göstərmək üçün pie chart-dan istifadə edin

```python
plt.figure(figsize=(4,4))
plt.pie(df['Type'].value_counts(),labels=df['Type'].value_counts().index,autopct='%1.1f%%')
plt.show()
```

9.Ən çox hansı filmlər ilk 10 film arasına girib?

```python
plt.figure(figsize=(20,10))
plt.bar(df['Title'].value_counts().head(10).index, df['Title'].value_counts().head(10))
plt.show()
```

10.İstifadəçi reytinqlərinə görə ən yüksək 10 film hansıdır? Nəticəni vizuallaşdırın

```python
plt.figure(figsize=(20,10))
plt.bar(df.groupby('Title')['Viewership Score'].max().sort_values(ascending=False).head(10).index,
        df.groupby('Title')['Viewership Score'].max().sort_values(ascending=False).head(10))
plt.show()
```

13.Netflix Exclusive filmlər daha yüksək reytinqə sahibdir, yoxsa digərləri?

14.Mən TV Show və Movie lər üçün aylıq trendi görmək istəyirəm.

```python
plt.figure(figsize=(12,5))
df['Month'] = df['As of'].dt.month

tv_show = df[df['Type'] == 'TV Show'].groupby('Month').count()
movie = df[df['Type'] == 'Movie'].groupby('Month').count()

plt.plot(tv_show.index, tv_show['Title'], label='TV Show')
plt.plot(movie.index, movie['Title'], label='Movie')

plt.legend()
plt.show()
```

15.Mən filmin Netflix-də yayınlanma tarixi ilə hər hansı bir korrelyasiyaya sahib olub olmadığını öyrənmək istəyirəm. Bunu həm düsturla, həm də vizualizasiya ilə göstərməlisiniz

```python
corr = df['Netflix Release Date'].corr(df['Viewership Score'])
corr
plt.scatter(df['Netflix Release Date'], df['Viewership Score'])
plt.show()
```

16.Yuxarıdakı xətaya əsasən, aşağıda təkcə illərə görə yeni bir korrelyasiya əmsalı hesablayın

```python
corr_year = df['Netflix Release Date'].dt.year.corr(df['Viewership Score'])
corr_year
```

17.Aylara görə ən məşhur Filmlər hansı olub?

```python
df['Month_Name'] = df['As of'].dt.month_name()
monthly_pop = df.groupby(['Month_Name', 'Title']).size().reset_index(name='Count')
monthly_pop = monthly_pop.sort_values('Count', ascending=False).drop_duplicates(['Month_Name'])
print("Aylara görə ən populyar olanlar:\n", monthly_pop)
```

18. 1/2/3/4 rübdə ən yüksək izləyici reytinqi olan 5 film hansı olub?

```python
df['Quarter'] = df['As of'].dt.quarter
quarter_movies = df[df['Type'] == 'Movie'].groupby(['Quarter', 'Title'])['Viewership Score'].sum().reset_index().sort_values(
    ['Quarter', 'Viewership Score'],ascending=[True, False])

top_5_movies = quarter_movies.groupby('Quarter').head(5)

top_5_movies
```

19.Yeni bir dashboard yaradın (15,15 ölçüsündə). Ən önəmli gördüyünüz 4 ədəd vizualı bu dashboard-da göstərin

```python
plt.figure(figsize=(15, 15))

plt.subplot(2, 2, 1)
plt.pie(df['Type'].value_counts(),labels=df['Type'].value_counts().index,autopct='%1.1f%%')

plt.subplot(2, 2, 2)
plt.bar(df['Title'].value_counts().head(10).index, df['Title'].value_counts().head(10))

plt.subplot(2, 2, 3)
plt.bar(df.groupby('Title')['Viewership Score'].max().sort_values(ascending=False).head(10).index,
        df.groupby('Title')['Viewership Score'].max().sort_values(ascending=False).head(10))
plt.subplot(2, 2, 4)
plt.scatter(df['Netflix Release Date'], df['Viewership Score'])

plt.show()
```
